# Dataset

```python
# create linearly seperable random dataset
import numpy as np
import matplotlib.pyplot as plt

in_n = np.random.uniform(0, 5, (10,2))
in_p = np.random.uniform(5, 10, (10,2))
features = np.concatenate((in_n,in_p))

labels = np.zeros(20)
labels[0:10] = 0
labels[10:20] = 1

#plt.scatter(features[:,0],features[:,1], c='g')
plt.scatter(in_p[:,0],in_p[:,1], c='g')
plt.scatter(in_n[:,0],in_n[:,1], c='r')
plt.title('Randomly Generated Dataset')
plt.xlabel('x1')
plt.ylabel('x2')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# init. weights
w = np.random.rand(2)
b = 0
print(w)
x1 = np.linspace(0,10)
x2 = -(w[0]*x1 + b)/w[1]
plt.plot(x1,x2,'-b')

for j in range(1000):
    r = np.random.randint(0,20)
    value = np.dot(w,features[r,:]) + b
    
    if labels[r]==1 and value < 0:
        plt.scatter(features[r,0],features[r,1], c='g')
        w = w + features[r,:]
        b = b + 1
        
    elif labels[r]==0 and value >= 0:
        plt.scatter(features[r,0],features[r,1], c='r')
        w = w - (features[r,:])
        b = b - 1
        
print(w)
x2 = -(w[0]*x1 + b)/w[1]
plt.plot(x1,x2,'-r')
```
